```python
def syndrome(R, L, G, n, m):
    """
    R: received word (list of length n),
    L: support of the goppa code (list of length n),
    G: goppa polynomial over GF(2^m),
    n: length of the vectors,
    m: dimension of the Galois group,
    return syndrome (polynomial over GF(2^m))
    """
    q = 2**m
    x = symbols('x')
    s = [0*n]
    s[i] = Poly(R[i]* invert(Poly(x - L[i] % Poly(G), domain = GF(q))) % Poly(G),
            domain =GF(q))
    S = Poly(s, domain = GF(q))
    return S
```

```python
def noisegen(R, n):
    """
    R: binary list,
    n: number of desired errors (n <= length(L)),
    returns a list with n errors
    """
    k = len(R) - 1
    nums = [randint(1, k) for _ in range(4)]
    R2 = list(R)
    for i in nums:
        R2[i] = (R2[i] + 1) % 2
    return R2

def hamming(R1, R2):
    diff = 0
    i = 0
    while i < len(R1):
        if R1[i] != R2[i]:
            diff += 1
        i += 1
    return diff
            
```

```python
R = [0,1,0,0,1,1,0,1,0]
R2 = noisegen(R, 4)
print(R)
print(R2)
print(hamming(R, R2))
```
