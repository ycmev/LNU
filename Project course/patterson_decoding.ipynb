```python
zech_16 = {'inf':0, 0:'inf', 1:4, 2:8, 3:14, 4:1, 5:10, 6:13, 7:9, 8:2, 9:7, 10:5, 11:12, 12:11, 13:6, 14:3}

    
def sum(i, j, m):
    """
    returns e such that alpha^i + alpha^j = alpha^e.
    """
    if i == j:
        s = z
    elif i == z:
        s = j
    elif j == z:
        s = i
    else:
        if i < j:
            s = (i + zech_16[j - i]) % (m)
        if j < i:
            s = (j + zech_16[i - j]) % (m)
    return s

def mult(i, j, m):
    """
    returns e such that alpha^i x alpha^j = alpha^e
    e.g.
    mult(inf, j) = inf,
    mult(j, i) = j + i (mod m)
    """
    if i == 'inf' or j == 'inf':
        return 'inf'
    else:
        return (i + j) % m

def inverse(i, m):
    """
    returns e such that alpha^e * alpha^i = 1
    """
    if i == 'inf':
        print('non invertible element')
    else:
        return (m - i) % m

def timesC(p, c, m):
    """
    returns list rep of alpha^c * p(z) where c is a constant
    """
    return [mult(p[i], c, m) for i in range(len(p))]

def timesZtoN(p, n, m):
    """
    returns list rep. of z^n * p(z) modulo z^4 + z + 1
    """
    if n == 0:
        return p
    if n == 1:
        p0, p1, p2, p3 = p
        return [p3, sum(p0, p3, m), p1, p2]
    else:
        return timesZtoN(timesZtoN(p, 1, m), n - 1, m)

def sumPoly(p, q, m):
    """
    p, q: two lists of the same lenght,
    m: positive integer
    returns the list representation of p + q modulo z^4 + z + 1
    """
    return [sum(p[i], q[i], m) for i in range(len(p))]


def timesPoly(p, q, m):
    """
    p: 4-length list,
    q: 4-length list,
    m: positive integer
    returns the list representation of p * q modulo z^4 + z + 1
    """
    """
    p(z) = a0 + a1 * z + a2 * z^2
    q(z) = b0 + b1 * z + b2 * z^2
    p(z) * q(z) = p(z) * b0 + p(z) * b1 * z + p(z) * b2 * z * z
    """
    r = ['inf'] * len(p)
    for i in range(len(q)):
        r = sumPoly(r, timesZtoN(timesC(p, q[i], m), i, m), m)
    return r

def multInv(p, m):
    """
    p: 4-length list
    m: the size of the F*
    return the representation of p^-1
    """

def dropRightInfs(g):
    """
    drops the infs from the right of the given list.
    e.g. [1, 0, 'inf', 'inf'] -> [1, 0]
    """
    i = len(g)
    while i > 0 and g[i-1] == 'inf':
        i -= 1
    return g[:i]

def degreeOf(g):
    """
    returns the degree of g, i.e., the last non-zero coefficient
    """
    i = len(g)
    while i > 0 and g[i - 1] == 'inf':
        i -= 1
    return i - 1

def zToNmod(g, m):
    """
    p,g: 4-length list
    m: the size of the multiplicative group F*_{2^m}
    return: p reduced modulo g
    """
    #1.start with g (the goppa polynomial). take its degree, which is the last nonzero entry in the list;
    t = degreeOf(g)
    #3. This index will determine the relation modulo g(z).
    pivot = g[t]
    coefs = list(g)
    coefs[t] = 'inf'
    for i in range(t):
        coefs[i] = mult(coefs[i], inverse(pivot, m), m)
    return coefs

```
