### Exercise 1. Probable Primes.

```python
def millerRabinWitnessQ (a ,n) :
    # Returns True if a is Millar - rabin witness for the compositness of n.
    # and False if it is maybe a prime
    # Find as many factors of 2 as possible in n -1
    k=n -1
    r =0
    while k %2==0:
        r=r +1
        k=k //2
    # Now n -1= k *2^ r and k is odd
    # print ("k =" ,k)
    # print ("r =" ,r)
    x= fast2Power (a ,k , n)
    if x ==1:
        return False
    for i in range (0 , r):
        if x ==n -1:
            return False
        x =( x **2) %n
    return True

def probablePrimeQ (n):
    # Runs Miller - Rabin primality test for 20 random values of a
    for i in range (0 ,20) :
        a= randint (2 ,n -1)
        if millerRabinWitnessQ (a ,n):
            return False
    return True

def findPrime ( lBound , uBound ) :
    while True :
        n= randint ( lBound , uBound )
        if n %2==1:
            if probablePrimeQ (n ):
                return n
```

```python
def generateRSA(b):
    l = 2**(b-1)
    u = 2**(b) - 1
    p = findPrime(l,u)
    q = findPrime(l,u)
    s = (p-1)*(q-1)
    while True:
        e = randint(1, s-1)
        g = extendedGCD(e,s)
        if g[0] == 1:
            d = g[1] % s
            break
    return [p,q,p*q,e,d]
```

```python
def findPrimitiveRoot(p):
    q = (p-1)//2
    while True:
        g = randint(2,p-2)
        if fast2Power(g,2,p) > 1 and fast2Power(g,q,p) == p-1:
            break
    return g
```

### Exercise 4. Pollard p-1.

```python
def factorPollard(N,a,m):
    b = a
    for i in range(1, m):
        b = fast2Power(b, i+1, N)
        d = extendedGCD(b - 1, N)[0]
        if d > 1 and d < N:
            return d
        elif d == N:
            return N
    return 1
```

### Exercise 6. Pollard Rho.

```python
def factorPollardRho(N,s=2):
    x = s
    y = x
    while True:
        x = (fast2Power(x,2,N) + 1) % N
        y = (fast2Power(fast2Power(y,2,N) + 1,2,N) + 1) % N
        d = extendedGCD(x - y, N)[0]
        if d > 1 and d < N:
            return d
```

### Exercise 8. Pollard Rho. Discrete logarithms.

```python
def solveLinear (a ,b ,m): # Solves the linear congruence equation ax=b( mod m)
    [d ,s ,t ]= extendedGCD (a ,m)
    print (d)
    solutions =[]
    if b% d ==0: #We have solutions if d divides b
        x0 =( s *( b // d ))%m
        mp =m // d
        for i in range ( d):
            solutions . append (( x0 +i* mp )%m )
    return solutions
```

```python
def pollardF(x, p, g, h):
    """
    x is a positive integer,
    p is a prime,
    g is a primitive root modulo p,
    h is an integer which we want to get its logarithm modulo p
    """
    if x < p//3:
        return g*x % p
    elif x > p//3 and x < 2*p//3:
        return fast2Power(x,2,p)
    else:
        return h * x  % p

def updateAlpha(a, x, p):
    if x < p//3:
        return (a + 1) % (p-1)
    elif x > p//3 and x < 2*p//3:
        return 2*a % (p-1)
    else:
        return a % (p-1)

def updateBeta(b, x, p):
    if x < p//3:
        return b % (p-1)
    elif x > p//3 and x < 2*p//3:
        return 2*b % (p-1)
    else:
        return (b + 1) % (p-1)
    
        
def discretePollardRho(g,h,p):
    x = pollardF(1,p,g,h)
    y = pollardF(x,p,g,h)
    alpha = 1
    beta = 0
    gamma = 2
    delta = 0
    while x != y:
        x = pollardF(x, p, g, h)
        alpha = updateAlpha(alpha, x, p)
        beta = updateBeta(beta, x, p)        
        y0 = pollardF(y, p, g, h)
        gamma0 = updateAlpha(gamma, y0,p)
        delta0 = updateBeta(delta, y0,p)
        y = pollardF(y0, p, g, h)
        gamma = updateAlpha(gamma0, y, p)
        delta = updateBeta(delta0, y,p)
    return solveLinear(alpha-gamma, beta-delta,p)
        
    
```
