```python
def divisor(a,b): #is a a divisor of b?
    if b%a==0:
        return True
    else:
        return False
```

```python
def findDivisors(listOfNumbers,m): #find divisors of m in listOfNumbers
    listOfDivisors=[]
    for i in listOfNumbers:
        if divisor(i,m):
            listOfDivisors.append(i)
    return listOfDivisors
```

```python
def fast2Power (a ,n ,m):
    res =1
    while n >0:
        if n %2==1: #If the bit is 1 multiply by the corresponding square
            res =( res *a )%m
        a =( a*a) %m
        n=n //2
    return res
```

```python
#extended euclidean algorithm
# Output [r,s,t] satisfying s*a+t*b=r=gcd(a,b)
def extendedGCD(a , b):
    r0 , r=a ,b
    s0 , s =1 ,0
    t0 , t =0 ,1
    while (r >0):
        tempr , temps , tempt =r ,s ,t
        q= r0 // r
        r ,s , t=r0 - q*r ,s0 -q*s ,t0 - q*t
        r0 , s0 , t0 = tempr , temps , tempt
    return [r0 ,s0 , t0 ]
```

```python
# Output the multiplicative inverse of a modulo p
def multInverse (a , p):
    result = extendedGCD (a , p)
    if result [0]!=1: # Error message if a and p are not relatively prime
        s=" Numbers needs to be relatively prime "
        return s
    inv = result [1]% p
    return inv
```

```python
# Finds the m- ary expansion of n
def getExpansion (n ,m):
    listOfDigits =[]
    while n >= m:
        digit =n%m
        listOfDigits . append ( digit )
        n =(n - digit ) // m
    listOfDigits.append(n)
    return listOfDigits
```

```python
# Chinese remainder theorem
def CRT (remainderList , modulusList ):
    M =1
    l =0
    for m in modulusList : # Computes the product of the moduli
        M=M *m
        l=l +1
    if l != len ( remainderList ):
        print (" Length of lists must be be the same ")
        return None
    x =0
    for i in range (0 , l):
        m= modulusList [i]
        Mi =M // m
        yi = multInverse ( Mi ,m)
        x =( x+ remainderList [i ]* Mi * yi )%M
    return x
```

# Part 3. Converting Algorithms

```python
s = "hello"
for i in s:
    print(ord(i))
```

```python
def textToInt (s):
    n =0
    k =0
    for i in s :
        n=n +ord( i) *(256** k)
        k=k +1
    return n
```

```python
def intToText (n):
    t= getExpansion(n ,256)
    myString =''
    for i in t :
        myString = myString + chr(i)
    return myString
```
