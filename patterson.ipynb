## Generate an error vector of a given size and weight

```python
import random as rd
```

```python
def randerror(n, t):
    """
    generates a list with at most t ones and remaining values zero
    :params n, t: positive integers
    :return: a list of size n with at most t ones randomly distributed and remaining values as zero.
    """
    e = [0]*n
    for i in range(t):
        k = rd.randint(0, n - 1)
        e[k] = 1
    return e
```
